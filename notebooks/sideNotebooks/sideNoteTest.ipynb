```python

import os
import pandas as pd
# Tạo thư mục lưu dữ liệu mẫu
os.makedirs("../../data/sample", exist_ok=True)
df = pd.read_csv(
    "../../data/raw/apartments_for_rent_classified_100K.csv",
    sep=";",
    encoding="cp1252",
    low_memory=False
)
# Lấy ngẫu nhiên 5.000 dòng từ dữ liệu gốc
df_sample = df.sample(n=1000, random_state=42)

# Lưu dữ liệu mẫu thành CSV
df_sample.to_csv(
    "../../data/sample/apartments_sample_1K.csv",
    sep=";",
    encoding="utf-8-sig",
    index=False
)

print("Kích thước dữ liệu gốc:", df.shape)
print("Kích thước dữ liệu mẫu:", df_sample.shape)
display(df_sample.head())
```
